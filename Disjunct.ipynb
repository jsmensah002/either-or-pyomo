```python
from pyomo.environ import *
from pyomo.gdp import Disjunct, Disjunction
```

```python
model = ConcreteModel()
```

```python
model.P = Var(within=NonNegativeReals, bounds = (0, 10000))
model.Q = Var(within=NonNegativeReals, bounds = (0, 10000))
model.Profit = Var(within=Reals, bounds = (-1000, 10000))
```

```python
model.MaximizeProfit = Objective(expr = model.Profit, sense = maximize)
```

```python
model.FeedConstraint = Constraint(expr = model.P + model.Q <=60)
model.DemandConstraint = Constraint(expr = model.P <=35)
```

```python
model.Reactor1 = Disjunct()
model.Reactor1.ProfitConstraint = Constraint(expr = model.Profit == 50*model.P + 40*model.Q)
model.Reactor1.SteamConstraint = Constraint(expr = 2*model.P + model.Q <=90)
```

```python
model.Reactor2 = Disjunct()
model.Reactor2.ProfitConstraint = Constraint(expr = model.Profit == 70*model.P + 40*model.Q)
model.Reactor2.SteamConstraint = Constraint(expr = 3*model.P + model.Q <=90)
```

```python
model.reactors = Disjunction(expr = [model.Reactor1, model.Reactor2], xor=True)
```

```python
TransformationFactory ("gdp.bigm").apply_to(model)
```

```python
optimizer = SolverFactory("appsi_highs")
optimizer.solve(model)
model.display()
```
